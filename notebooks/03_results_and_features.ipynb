```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'results').exists():
    ROOT = ROOT.parent
RESULTS = ROOT / 'results'

validation = pd.read_csv(RESULTS / 'model_validation_comparison.csv')
final_test = pd.read_csv(RESULTS / 'official_test_results.csv')
features = pd.read_csv(RESULTS / 'selected_features.csv')

print('=== Validation Comparison (Detection & Hardware Resource Metrics) ===')
display(validation)

print('\n=== Independent Official Test Result (Final Evaluation) ===')
display(final_test)

print('\n=== Hardware & System Resource Benchmark Summary ===')
res_cols = ['model', 'train_time_sec', 'train_cpu_percent', 'ram_usage_mb', 'train_ram_delta_mb', 'inference_time_sec', 'inference_latency_ms', 'throughput_samples_per_sec', 'inference_cpu_percent']
available_cols = [c for c in res_cols if c in validation.columns]
if available_cols:
    display(validation[available_cols])

print('\n=== Thirty Selected Model Inputs ===')
display(features)
```
